```python
# 1. Environment check + imports
import os
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.layers import TextVectorization
from sklearn.model_selection import train_test_split

print("TF version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))
```

```python
# 2. Load dataset (adjust path if your file is elsewhere)
df = pd.read_csv(r"C:\Users\PUNITH KUMAR\Desktop\kaggle\train.csv")
print("Loaded df shape:", df.shape)
df.head(2)
```

```python
# 3. Prepare texts and labels
texts = df["comment_text"].astype(str).values
labels = df[["toxic","severe_toxic","obscene","threat","insult","identity_hate"]].astype(int).values
print("Example text:", texts[0][:120])
print("Labels shape:", labels.shape)
```

```python
# 4. Train/validation split (80/20)
train_texts, val_texts, y_train, y_val = train_test_split(
    texts, labels, test_size=0.2, random_state=42, stratify=None
)
print("Train size:", len(train_texts), "Val size:", len(val_texts))
```

```python
# 5. Vectorizer: reduce sequence length to speed up training
MAX_FEATURES = 50000    # vocab size (tune down to 20000 if memory is tight)
max_length = 300        # <- much faster than 1800

vectorizer = TextVectorization(
    max_tokens=MAX_FEATURES,
    output_sequence_length=max_length,
    output_mode='int'
)

# adapt on training texts (this builds the vocabulary)
vectorizer.adapt(train_texts)
vocab_size = len(vectorizer.get_vocabulary())
print("Vocab size:", vocab_size)
```

```python
# 6. Vectorize and create datasets (increase batch_size if you have GPU/RAM)
batch_size = 64   # try 128 if GPU memory permits
AUTOTUNE = tf.data.AUTOTUNE

# vectorize to integer sequences
vectorized_train = vectorizer(np.array(train_texts.tolist()))
vectorized_val   = vectorizer(np.array(val_texts.tolist()))

# convert labels to numpy arrays
y_train = np.array(y_train)
y_val   = np.array(y_val)

# build tf.data pipelines
train_ds = tf.data.Dataset.from_tensor_slices((vectorized_train, y_train))
train_ds = train_ds.cache()
train_ds = train_ds.shuffle(160000)
train_ds = train_ds.batch(batch_size)
train_ds = train_ds.prefetch(AUTOTUNE)

val_ds = tf.data.Dataset.from_tensor_slices((vectorized_val, y_val))
val_ds = val_ds.batch(batch_size).prefetch(AUTOTUNE)

print("Train batches (one epoch):", tf.data.experimental.cardinality(train_ds).numpy())
print("Val batches:", tf.data.experimental.cardinality(val_ds).numpy())
```

```python
from tensorflow.keras import layers, Sequential

model = Sequential([
    layers.Embedding(vocab_size, 128),
    layers.Bidirectional(layers.GRU(64)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(16, activation='relu'),
    layers.Dense(6, activation='sigmoid')
])
```

```python
import tensorflow as tf
print("TF version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))
```

```python
# ====== FULL TRAINING BLOCK (RUN THIS DIRECTLY) ======

import tensorflow as tf
from tensorflow.keras import layers, Sequential
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau

# ---- MODEL ----
model = Sequential([
    layers.Embedding(vocab_size, 128),
    layers.Bidirectional(layers.GRU(64)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(16, activation='relu'),
    layers.Dense(6, activation='sigmoid')    # multi-label output
])

# ---- COMPILE (MUST BE DONE BEFORE FIT) ----
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("MODEL COMPILED SUCCESSFULLY\n")
model.summary()

# ---- CALLBACKS ----
early_stop = EarlyStopping(
    monitor='val_loss', patience=3, restore_best_weights=True
)

checkpoint = ModelCheckpoint(
    'best_toxic_model.keras',   # MUST END WITH .keras
    monitor='val_loss',
    save_best_only=True
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=2,
    min_lr=1e-6
)

callbacks = [early_stop, checkpoint, reduce_lr]

# ---- TRAIN ----
epochs = 5

history = model.fit(
    train_ds,
    epochs=epochs,
    validation_data=val_ds,
    callbacks=callbacks
)

print("TRAINING COMPLETE!")
```

## 4. Evaluate Model

```python
from tensorflow.keras.metrics import Precision, Recall, CategoricalAccuracy
```

```python
pre = Precision()
re = Recall()
acc = CategoricalAccuracy()
```

```python
for batch in test.as_numpy_iterator(): 
    # Unpack the batch 
    X_true, y_true = batch
    # Make a prediction 
    yhat = model.predict(X_true)
    
    # Flatten the predictions
    y_true = y_true.flatten()
    yhat = yhat.flatten()
    
    pre.update_state(y_true, yhat)
    re.update_state(y_true, yhat)
    acc.update_state(y_true, yhat)
```

```python
print(f'Precision: {pre.result().numpy()}, Recall:{re.result().numpy()}, Accuracy:{acc.result().numpy()}')
```

```python
def score_comment(comment):
    vectorized_comment = vectorizer([comment])
    results = model.predict(vectorized_comment)
    
    text = ''
    for idx, col in enumerate(df.columns[2:]):
        text += '{}: {}\n'.format(col, results[0][idx]>0.5)
    
    return text
```

```python
interface = gr.Interface(fn=score_comment, 
                         inputs=gr.inputs.Textbox(lines=2, placeholder='Comment to score'),
                        outputs='text')
```
